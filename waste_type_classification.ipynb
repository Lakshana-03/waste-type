```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
import joblib
```

```python
data=pd.read_csv("/content/waste_classification_dataset_500.csv")
print(data.head())
```

```python
x=data.drop("WasteType",axis=1)
y=data["WasteType"]
```

```python
encoder=LabelEncoder()
y=encoder.fit_transform(y)
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
lr=LogisticRegression()
lr.fit(x_train,y_train)
lr_pred=lr.predict(x_test)
lr_pred
```

```python
print("Logistic Regression Accuracy")
print(accuracy_score(y_test,lr_pred))
print("\nClassification Report")
print(classification_report(y_test,lr_pred))
print("\nConfusion Matrix")
print(confusion_matrix(y_test,lr_pred))
joblib.dump(lr,'Waste Type Classification_Logistic Regression.pkl')
```

```python
dt=DecisionTreeClassifier(random_state=42)
dt.fit(x_train,y_train)
dt_pred=dt.predict(x_test)
```

```python
print("Decision Tree Accuracy")
print(accuracy_score(y_test,dt_pred))
print("\nClassification Report")
print(classification_report(y_test,dt_pred))
print("\nConfusion Matrix")
print(confusion_matrix(y_test,dt_pred))
```

```python
dt_accuracy = accuracy_score(y_test, dt_pred)
lr_accuracy = accuracy_score(y_test, lr_pred)

print("Decision Tree :", dt_accuracy)

print("Logistic Regression :", lr_accuracy)

if dt_accuracy > lr_accuracy:
    print("Decision Tree performed better.")
elif lr_accuracy > dt_accuracy:
    print("Logistic Regression performed better.")
else:
    print("Both models performed equally.")
```

```python
new_sample = [[90,65,3,0,1]]

prediction = dt.predict(new_sample)

print("Predicted Waste Type:")

print(encoder.inverse_transform(prediction))
```
